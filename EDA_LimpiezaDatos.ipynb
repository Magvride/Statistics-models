### Reordenamiento cronologico y detección de valores nulos

```python
def EDA_clean_sort(df):

    filas_iniciales = len(df)

    #Reescribir la fecha como datetime

    df['Date'] = pd.to_datetime(
        df[['YY','MM','DD','hh','mm']]
        .rename(columns={
            'YY':'year',
            'MM':'month',
            'DD':'day',
            'hh':'hour',
            'mm':'minute'
        })
    )

    df = df[['Date','WDIR','WSPD','GST','WVHT','DPD','APD','MWD','PRES','ATMP','WTMP','DEWP','VIS','PTDY','TIDE']]
    
    #Volver los valores en NaN y eliminar columnas completamente vacias

    df = df.replace([99,999,9999], np.nan)
    df = df.dropna(axis=1, how='all')

    #Ordenar por datetime
    df = df.sort_values(['Date']).reset_index(drop=True)

    #Eliminar filas con valores NaN(Los reportes son cada 30 min)

    df = df.dropna().reset_index(drop=True)

    filas_finales = len(df)

    print("Resultados: " )
    print(f"Filas eliminadas: {filas_iniciales - filas_finales}")
    print(f"Filas restantes: {filas_finales}\n")

    return df
```

```python
df23 = EDA_clean_sort(df23)
df24 = EDA_clean_sort(df24)
df25 = EDA_clean_sort(df25)
df2601 = EDA_clean_sort(df2601)
df2604 = EDA_clean_sort(df2604)
df2605 = EDA_clean_sort(df2605)

df23
```
