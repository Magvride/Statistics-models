```python
import pandas as pd
import numpy as np

columnas = ['YY', 'MM', 'DD', 'hh', 'mm', 'WDIR', 'WSPD', 'GST', 'WVHT',
            'DPD', 'APD', 'MWD', 'PRES', 'ATMP', 'WTMP', 'DEWP', 'VIS', 'TIDE']

df = pd.read_fwf('datos_historicos_2023', skiprows=2, names=columnas)
df.replace([99.0, 99.00, 999, 999.0], np.nan, inplace=True)
df.head()
```

```python
df = df[['YY', 'MM', 'DD', 'hh', 'mm', 'WVHT', 'WSPD', 'WTMP']]

df['datetime'] = pd.to_datetime(
    df[['YY', 'MM', 'DD', 'hh', 'mm']].rename(columns={
        'YY': 'year',
        'MM': 'month',
        'DD': 'day',
        'hh': 'hour',
        'mm': 'minute' 
    })
)

df = df.rename(columns={
    'WVHT': 'Hs',
    'WSPD': 'Wind_speed',
    'WTMP': 'sst'
})

df = df[['datetime', 'Hs', 'Wind_speed', 'sst']]

df.head()
```

```python
from statsmodels.tsa.stattools import adfuller

variables = ['Wind_speed', 'sst', 'Hs']

for var in variables:
    
    # Eliminar NaN
    serie = df[var].dropna()
    
    # Aplicar ADF test
    resultado = adfuller(serie)
    
    print(f'\nADF Test para {var}')
    print(f'ADF Statistic: {resultado[0]}')
    print(f'p-value: {resultado[1]}')
    
    if resultado[1] < 0.05:
        print('La serie es estacionaria')
    else:
        print('La serie NO es estacionaria')
```
